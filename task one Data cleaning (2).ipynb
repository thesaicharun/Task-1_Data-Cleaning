```python
# =========================
# Final Data Cleaning Pass
# =========================

# 1. Remove completely empty columns
empty_cols = df.columns[df.isnull().all()]
df.drop(columns=empty_cols, inplace=True)

# 2. Handle missing values in object columns
object_cols = df.select_dtypes(include="object").columns

for col in object_cols:
    df[col] = df[col].fillna("Unknown")

# 3. Handle missing values in numeric columns
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# 4. Handle missing dates
if "date_added" in df.columns:
    df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

    # Replace missing dates with most frequent date
    df["date_added"] = df["date_added"].fillna(
        df["date_added"].mode()[0]
    )

# 5. Remove duplicate rows
df.drop_duplicates(inplace=True)

# 6. Standardize text columns
for col in object_cols:
    df[col] = df[col].astype(str).str.strip()

# 7. Clean column names
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

# =========================
# Final Validation
# =========================

print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

df.info()
```
